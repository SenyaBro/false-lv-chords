```python
from pathlib import Path

import numpy as np
import pandas as pd

pd.set_option("display.max_columns", 200)
pd.set_option("display.width", 200)

PROJECT_ROOT = Path(r"C:\Users\Ars\projects\university\lab_urfu_2026\false-lv-chords")

DATA_PATH = PROJECT_ROOT / "data" / "interim" / "regression_working_dataset_strict.xlsx"

OUTPUT_DIR = PROJECT_ROOT / "output"
TABLES_DIR = OUTPUT_DIR / "tables"
INTERIM_DIR = PROJECT_ROOT / "data" / "interim"

for path in [OUTPUT_DIR, TABLES_DIR, INTERIM_DIR]:
    path.mkdir(parents=True, exist_ok=True)

df = pd.read_excel(DATA_PATH)

print("Shape:", df.shape)
display(df.head())
```

```python
TRANSVERSE_COLS = [
    "lv_false_tendon_mid_transverse_count",
    "lv_false_tendon_mid_and_basal_transverse_count",
    "lv_false_tendon_basal_transverse_count",
]

OBLIQUE_COLS = [
    "lv_false_tendon_mid_oblique_count",
    "lv_false_tendon_basal_oblique_count",
    "lv_false_tendon_mid_to_basal_oblique_count",
    "lv_false_tendon_mid_to_apical_oblique_count",
    "lv_false_tendon_basal_to_apical_oblique_count",
]

LONG_OBLIQUE_COLS = [
    "lv_false_tendon_mid_to_basal_oblique_count",
    "lv_false_tendon_mid_to_apical_oblique_count",
    "lv_false_tendon_basal_to_apical_oblique_count",
]

ALL_CHORD_TYPE_COLS = TRANSVERSE_COLS + OBLIQUE_COLS

for col in ALL_CHORD_TYPE_COLS:
    if col not in df.columns:
        raise ValueError(f"Missing column: {col}")

# 1. Поперечный tethering-фенотип
df["transverse_load"] = df[TRANSVERSE_COLS].sum(axis=1, min_count=1)
df["transverse_any"] = (df["transverse_load"] > 0).astype("Int64")

# 2. Косой фенотип
df["oblique_load"] = df[OBLIQUE_COLS].sum(axis=1, min_count=1)
df["oblique_any"] = (df["oblique_load"] > 0).astype("Int64")

# 3. Длинные косые мосты
df["long_oblique_bridge_load"] = df[LONG_OBLIQUE_COLS].sum(axis=1, min_count=1)
df["long_oblique_bridge_any"] = (df["long_oblique_bridge_load"] > 0).astype("Int64")

# 4. Количество разных типов хорд
df["chord_type_diversity"] = (df[ALL_CHORD_TYPE_COLS] > 0).sum(axis=1)

# 5. Упрощенная сложность сети
def classify_complexity(x):
    if pd.isna(x):
        return np.nan
    if x == 0:
        return "0_no_selected_types"
    if x == 1:
        return "1_single_type"
    if x == 2:
        return "2_two_types"
    return "3_complex_network"

df["chord_type_complexity_class"] = df["chord_type_diversity"].apply(classify_complexity)

# 6. Фенотип: поперечные против только косых
def classify_mechanical_phenotype(row):
    transverse = row["transverse_any"]
    oblique = row["oblique_any"]
    
    if pd.isna(transverse) or pd.isna(oblique):
        return np.nan
    
    if transverse == 1:
        return "Transverse-Involved"
    if transverse == 0 and oblique == 1:
        return "Oblique-Only"
    return "No-selected-chord-types"

df["mechanical_chord_phenotype"] = df.apply(classify_mechanical_phenotype, axis=1)

# 7. Смешанный индекс tethering
# Поперечные хорды получают больший вес, потому что механически они ближе к распоркам
df["tethering_index"] = (
    2.0 * df["transverse_load"].fillna(0)
    + 1.0 * df["long_oblique_bridge_load"].fillna(0)
)

MECHANISTIC_FEATURES = [
    "transverse_any",
    "transverse_load",
    "oblique_any",
    "oblique_load",
    "long_oblique_bridge_any",
    "long_oblique_bridge_load",
    "chord_type_diversity",
    "chord_type_complexity_class",
    "mechanical_chord_phenotype",
    "tethering_index",
]

display(df[MECHANISTIC_FEATURES].head(20))
```

```python
summary_rows = []

for col in MECHANISTIC_FEATURES:
    s = df[col]
    
    if pd.api.types.is_numeric_dtype(s):
        summary_rows.append({
            "feature": col,
            "type": "numeric",
            "n_non_missing": int(s.notna().sum()),
            "n_missing": int(s.isna().sum()),
            "unique_values": int(s.nunique(dropna=True)),
            "mean": s.mean(),
            "sd": s.std(),
            "median": s.median(),
            "min": s.min(),
            "max": s.max(),
            "n_nonzero": int((s > 0).sum()) if s.notna().sum() > 0 else np.nan,
        })
    else:
        summary_rows.append({
            "feature": col,
            "type": "categorical",
            "n_non_missing": int(s.notna().sum()),
            "n_missing": int(s.isna().sum()),
            "unique_values": int(s.nunique(dropna=True)),
            "mean": np.nan,
            "sd": np.nan,
            "median": np.nan,
            "min": np.nan,
            "max": np.nan,
            "n_nonzero": np.nan,
        })

mechanistic_summary = pd.DataFrame(summary_rows)

display(mechanistic_summary)

for col in MECHANISTIC_FEATURES:
    print("\n", "=" * 100)
    print(col)
    display(df[col].value_counts(dropna=False).to_frame("n"))

mechanistic_summary.to_excel(TABLES_DIR / "01_mechanistic_feature_summary.xlsx", index=False)
```

```python
TARGETS = [
    "echo_lv_strain_plus_cv_pct",
    "echo_lv_strain_rate_cv_pct",
    "dicor_ssou_cv_pct",
    "dicor_sm_kdtr_cv_pct",
    "echo_lv_sphericity_index_end_systolic_ratio",
    "echo_lv_end_systolic_diameter_mm",
]

MECHANISTIC_PREDICTORS = [
    "transverse_any",
    "transverse_load",
    "oblique_any",
    "oblique_load",
    "long_oblique_bridge_any",
    "long_oblique_bridge_load",
    "chord_type_diversity",
    "tethering_index",
    "mechanical_chord_phenotype",
    "chord_type_complexity_class",
]

availability_rows = []

for target in TARGETS:
    for predictor in MECHANISTIC_PREDICTORS:
        if target not in df.columns or predictor not in df.columns:
            continue
        
        d = df[[target, predictor]].dropna().copy()
        
        if len(d) == 0:
            continue
        
        availability_rows.append({
            "target": target,
            "predictor": predictor,
            "n_complete": len(d),
            "predictor_unique_values": int(d[predictor].nunique(dropna=True)),
            "target_mean": d[target].mean(),
            "target_sd": d[target].std(),
        })

availability = pd.DataFrame(availability_rows)

availability = availability.sort_values(
    ["target", "n_complete", "predictor_unique_values"],
    ascending=[True, False, False]
)

display(availability)

availability.to_excel(TABLES_DIR / "01_mechanistic_pairwise_availability.xlsx", index=False)
```
