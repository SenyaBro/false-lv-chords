```python
from pathlib import Path
import warnings

import numpy as np
import pandas as pd

import scipy.stats as stats
import statsmodels.formula.api as smf

import seaborn as sns
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", 200)
pd.set_option("display.width", 200)

PROJECT_ROOT = Path(r"C:\Users\Ars\projects\university\lab_urfu_2026\false-lv-chords")

DATA_PATH = PROJECT_ROOT / "data" / "interim" / "regression_mechanistic_dataset.xlsx"

OUTPUT_DIR = PROJECT_ROOT / "output"
FIGURES_DIR = OUTPUT_DIR / "figures"
TABLES_DIR = OUTPUT_DIR / "tables"

FIGURES_DIR.mkdir(parents=True, exist_ok=True)
TABLES_DIR.mkdir(parents=True, exist_ok=True)

df = pd.read_excel(DATA_PATH)

print("Shape:", df.shape)
display(df.head())
```

```python
TARGETS = [
    "echo_lv_strain_plus_cv_pct",
    "echo_lv_strain_rate_cv_pct",
    "dicor_ssou_cv_pct",
    "dicor_sm_kdtr_cv_pct",
    "echo_lv_sphericity_index_end_systolic_ratio",
    "echo_lv_end_systolic_diameter_mm",
]

PHENOTYPE_COL = "mechanical_chord_phenotype"

KEEP_GROUPS = [
    "Oblique-Only",
    "Transverse-Involved",
]

TARGETS = [c for c in TARGETS if c in df.columns]

if PHENOTYPE_COL not in df.columns:
    raise ValueError(f"Column not found: {PHENOTYPE_COL}")

df_pheno = df[df[PHENOTYPE_COL].isin(KEEP_GROUPS)].copy()

df_pheno["phenotype_binary"] = np.where(
    df_pheno[PHENOTYPE_COL] == "Transverse-Involved",
    1,
    0,
)

df_pheno["phenotype_label"] = pd.Categorical(
    df_pheno[PHENOTYPE_COL],
    categories=["Oblique-Only", "Transverse-Involved"],
    ordered=True,
)

print("Filtered shape:", df_pheno.shape)
display(df_pheno[PHENOTYPE_COL].value_counts(dropna=False).to_frame("n"))

display(df_pheno[[PHENOTYPE_COL, "phenotype_binary", "phenotype_label"]].head())
```

```python
def cohen_d_independent(y0, y1):
    """
    Cohen's d для двух независимых групп.
    d = mean(y1) - mean(y0) / pooled SD
    """
    y0 = pd.Series(y0).dropna().astype(float)
    y1 = pd.Series(y1).dropna().astype(float)

    n0 = len(y0)
    n1 = len(y1)

    if n0 < 2 or n1 < 2:
        return np.nan

    s0 = y0.std(ddof=1)
    s1 = y1.std(ddof=1)

    pooled_sd = np.sqrt(((n0 - 1) * s0**2 + (n1 - 1) * s1**2) / (n0 + n1 - 2))

    if pooled_sd == 0:
        return np.nan

    return (y1.mean() - y0.mean()) / pooled_sd


def hedges_g_independent(y0, y1):
    """
    Hedges' g — поправка Cohen's d для малых выборок.
    """
    y0 = pd.Series(y0).dropna().astype(float)
    y1 = pd.Series(y1).dropna().astype(float)

    d = cohen_d_independent(y0, y1)
    n0 = len(y0)
    n1 = len(y1)

    if pd.isna(d) or (n0 + n1) <= 3:
        return np.nan

    correction = 1 - (3 / (4 * (n0 + n1) - 9))
    return d * correction


def bootstrap_mean_difference(y0, y1, n_boot=10000, random_state=42):
    """
    Bootstrap CI для разницы средних:
    Transverse-Involved minus Oblique-Only.
    """
    rng = np.random.default_rng(random_state)

    y0 = np.asarray(pd.Series(y0).dropna().astype(float))
    y1 = np.asarray(pd.Series(y1).dropna().astype(float))

    if len(y0) < 2 or len(y1) < 2:
        return np.nan, np.nan, np.nan

    diffs = []

    for _ in range(n_boot):
        sample0 = rng.choice(y0, size=len(y0), replace=True)
        sample1 = rng.choice(y1, size=len(y1), replace=True)
        diffs.append(sample1.mean() - sample0.mean())

    diffs = np.asarray(diffs)

    return (
        float(np.mean(diffs)),
        float(np.percentile(diffs, 2.5)),
        float(np.percentile(diffs, 97.5)),
    )


def cliffs_delta(y0, y1):
    """
    Cliff's delta:
    положительное значение означает, что y1 чаще больше y0.
    """
    y0 = np.asarray(pd.Series(y0).dropna().astype(float))
    y1 = np.asarray(pd.Series(y1).dropna().astype(float))

    if len(y0) == 0 or len(y1) == 0:
        return np.nan

    greater = 0
    lower = 0

    for value1 in y1:
        greater += np.sum(value1 > y0)
        lower += np.sum(value1 < y0)

    return (greater - lower) / (len(y0) * len(y1))
```

```python
def analyze_oblique_vs_transverse(df, target, phenotype_col="phenotype_label", min_group_n=3):
    d = df[[target, phenotype_col, "phenotype_binary"]].dropna().copy()

    d[target] = pd.to_numeric(d[target], errors="coerce")
    d = d.dropna(subset=[target, phenotype_col, "phenotype_binary"]).copy()

    group_counts = d[phenotype_col].value_counts()

    if len(group_counts) < 2:
        return None

    if group_counts.min() < min_group_n:
        return None

    y_oblique = d.loc[d[phenotype_col] == "Oblique-Only", target]
    y_transverse = d.loc[d[phenotype_col] == "Transverse-Involved", target]

    # OLS:
    # phenotype_binary = 0: Oblique-Only
    # phenotype_binary = 1: Transverse-Involved
    # coefficient = Transverse-Involved minus Oblique-Only
    formula = f'Q("{target}") ~ phenotype_binary'
    model = smf.ols(formula, data=d).fit()
    robust = model.get_robustcov_results(cov_type="HC3")

    robust_params = pd.Series(robust.params, index=model.params.index)
    robust_pvalues = pd.Series(robust.pvalues, index=model.params.index)
    robust_conf = pd.DataFrame(
        robust.conf_int(),
        index=model.params.index,
        columns=["ci_low_hc3", "ci_high_hc3"]
    )

    try:
        welch = stats.ttest_ind(y_oblique, y_transverse, equal_var=False, nan_policy="omit")
        welch_p = float(welch.pvalue)
    except Exception:
        welch_p = np.nan

    try:
        mw_p = float(stats.mannwhitneyu(y_oblique, y_transverse, alternative="two-sided").pvalue)
    except Exception:
        mw_p = np.nan

    boot_mean_diff, boot_ci_low, boot_ci_high = bootstrap_mean_difference(
        y_oblique,
        y_transverse,
        n_boot=10000,
        random_state=42,
    )

    result = {
        "target": target,
        "n_total": int(len(d)),

        "n_oblique_only": int(len(y_oblique)),
        "n_transverse_involved": int(len(y_transverse)),

        "mean_oblique_only": float(y_oblique.mean()),
        "sd_oblique_only": float(y_oblique.std(ddof=1)),
        "median_oblique_only": float(y_oblique.median()),
        "iqr_oblique_only": float(y_oblique.quantile(0.75) - y_oblique.quantile(0.25)),

        "mean_transverse_involved": float(y_transverse.mean()),
        "sd_transverse_involved": float(y_transverse.std(ddof=1)),
        "median_transverse_involved": float(y_transverse.median()),
        "iqr_transverse_involved": float(y_transverse.quantile(0.75) - y_transverse.quantile(0.25)),

        "mean_difference_transverse_minus_oblique": float(y_transverse.mean() - y_oblique.mean()),
        "median_difference_transverse_minus_oblique": float(y_transverse.median() - y_oblique.median()),

        "ols_intercept_oblique_mean": float(model.params["Intercept"]),
        "ols_beta_transverse_minus_oblique": float(model.params["phenotype_binary"]),
        "ols_beta_hc3": float(robust_params["phenotype_binary"]),
        "ols_beta_p_classic": float(model.pvalues["phenotype_binary"]),
        "ols_beta_p_hc3": float(robust_pvalues["phenotype_binary"]),
        "ols_beta_ci_low_hc3": float(robust_conf.loc["phenotype_binary", "ci_low_hc3"]),
        "ols_beta_ci_high_hc3": float(robust_conf.loc["phenotype_binary", "ci_high_hc3"]),

        "r2": float(model.rsquared),
        "adj_r2": float(model.rsquared_adj),
        "aic": float(model.aic),
        "bic": float(model.bic),

        "welch_ttest_p": welch_p,
        "mann_whitney_p": mw_p,

        "cohen_d_transverse_minus_oblique": float(cohen_d_independent(y_oblique, y_transverse)),
        "hedges_g_transverse_minus_oblique": float(hedges_g_independent(y_oblique, y_transverse)),
        "cliffs_delta_transverse_minus_oblique": float(cliffs_delta(y_oblique, y_transverse)),

        "bootstrap_mean_difference": boot_mean_diff,
        "bootstrap_ci_low_95": boot_ci_low,
        "bootstrap_ci_high_95": boot_ci_high,
    }

    return result, d, model
```

```python
results = []
model_objects = {}

for target in TARGETS:
    out = analyze_oblique_vs_transverse(
        df=df_pheno,
        target=target,
        phenotype_col="phenotype_label",
        min_group_n=3,
    )

    if out is None:
        print(f"Skipped: {target}")
        continue

    result, d_target, model = out

    results.append(result)
    model_objects[target] = {
        "data": d_target,
        "model": model,
    }

results_df = pd.DataFrame(results)

if results_df.empty:
    raise ValueError("No valid Oblique-Only vs Transverse-Involved models were fitted.")

results_df = results_df.sort_values(
    ["adj_r2", "n_total"],
    ascending=[False, False],
)

display(results_df)

results_df.to_excel(TABLES_DIR / "03_oblique_vs_transverse_results.xlsx", index=False)

print("Saved:")
print(TABLES_DIR / "03_oblique_vs_transverse_results.xlsx")
```

```python
group_summary_rows = []

for target in TARGETS:
    if target not in df_pheno.columns:
        continue

    d = df_pheno[[target, "phenotype_label"]].dropna().copy()
    d[target] = pd.to_numeric(d[target], errors="coerce")
    d = d.dropna(subset=[target])

    if d.empty:
        continue

    summary = (
        d.groupby("phenotype_label", observed=True)[target]
        .agg(["count", "mean", "std", "median", "min", "max"])
        .reset_index()
    )

    summary.insert(0, "target", target)
    group_summary_rows.append(summary)

group_summary_df = pd.concat(group_summary_rows, ignore_index=True)

display(group_summary_df)

group_summary_df.to_excel(TABLES_DIR / "03_oblique_vs_transverse_group_summary.xlsx", index=False)

print("Saved:")
print(TABLES_DIR / "03_oblique_vs_transverse_group_summary.xlsx")
```

```python
sns.set_theme(style="white", context="paper", font_scale=1.15)

def format_axes(ax):
    ax.set_facecolor("white")
    ax.grid(False)

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    ax.spines["left"].set_color("black")
    ax.spines["bottom"].set_color("black")
    ax.spines["left"].set_linewidth(1.1)
    ax.spines["bottom"].set_linewidth(1.1)

    ax.tick_params(axis="both", colors="black", width=1.0, length=4)
    ax.xaxis.label.set_color("black")
    ax.yaxis.label.set_color("black")


def plot_oblique_vs_transverse_box_strip(df, target, result_row, save_path):
    d = df[[target, "phenotype_label"]].dropna().copy()
    d[target] = pd.to_numeric(d[target], errors="coerce")
    d = d.dropna(subset=[target])

    order = ["Oblique-Only", "Transverse-Involved"]

    fig, ax = plt.subplots(figsize=(5.2, 4.4), dpi=300)

    sns.boxplot(
        data=d,
        x="phenotype_label",
        y=target,
        order=order,
        width=0.55,
        showfliers=False,
        ax=ax,
        boxprops={"facecolor": "white", "edgecolor": "black", "linewidth": 1.1},
        whiskerprops={"color": "black", "linewidth": 1.1},
        capprops={"color": "black", "linewidth": 1.1},
        medianprops={"color": "black", "linewidth": 1.2},
    )

    sns.stripplot(
        data=d,
        x="phenotype_label",
        y=target,
        order=order,
        ax=ax,
        color="black",
        size=4,
        jitter=0.18,
        alpha=0.65,
    )

    ax.set_xlabel("Mechanical chord phenotype")
    ax.set_ylabel(target)

    text = (
        f"n = {int(result_row['n_total'])}\n"
        f"R² = {result_row['r2']:.3f}\n"
        f"adj. R² = {result_row['adj_r2']:.3f}\n"
        f"β = {result_row['ols_beta_transverse_minus_oblique']:.2f}\n"
        f"p(HC3) = {result_row['ols_beta_p_hc3']:.3f}\n"
        f"Hedges g = {result_row['hedges_g_transverse_minus_oblique']:.2f}"
    )

    ax.text(
        0.03,
        0.97,
        text,
        transform=ax.transAxes,
        ha="left",
        va="top",
        fontsize=8.5,
        color="black",
        bbox={
            "facecolor": "white",
            "edgecolor": "black",
            "linewidth": 0.6,
            "alpha": 0.9,
        },
    )

    format_axes(ax)

    plt.xticks(rotation=15, ha="right")
    fig.tight_layout()
    fig.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.close(fig)


def plot_oblique_vs_transverse_mean_ci(result_row, save_path):
    labels = ["Oblique-Only", "Transverse-Involved"]

    means = [
        result_row["mean_oblique_only"],
        result_row["mean_transverse_involved"],
    ]

    sds = [
        result_row["sd_oblique_only"],
        result_row["sd_transverse_involved"],
    ]

    ns = [
        result_row["n_oblique_only"],
        result_row["n_transverse_involved"],
    ]

    sems = [
        sds[0] / np.sqrt(ns[0]) if ns[0] > 0 else np.nan,
        sds[1] / np.sqrt(ns[1]) if ns[1] > 0 else np.nan,
    ]

    ci95 = [1.96 * sems[0], 1.96 * sems[1]]

    fig, ax = plt.subplots(figsize=(4.8, 4.2), dpi=300)

    x = np.arange(len(labels))

    ax.errorbar(
        x,
        means,
        yerr=ci95,
        fmt="o",
        color="black",
        ecolor="black",
        elinewidth=1.1,
        capsize=4,
        markersize=5,
    )

    ax.set_xticks(x)
    ax.set_xticklabels(labels, rotation=15, ha="right")

    ax.set_xlabel("Mechanical chord phenotype")
    ax.set_ylabel(result_row["target"])

    text = (
        f"Δ mean = {result_row['mean_difference_transverse_minus_oblique']:.2f}\n"
        f"Bootstrap 95% CI:\n"
        f"[{result_row['bootstrap_ci_low_95']:.2f}; {result_row['bootstrap_ci_high_95']:.2f}]"
    )

    ax.text(
        0.03,
        0.97,
        text,
        transform=ax.transAxes,
        ha="left",
        va="top",
        fontsize=8.5,
        color="black",
        bbox={
            "facecolor": "white",
            "edgecolor": "black",
            "linewidth": 0.6,
            "alpha": 0.9,
        },
    )

    format_axes(ax)

    fig.tight_layout()
    fig.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.close(fig)
```

```python
for _, row in results_df.iterrows():
    target = row["target"]
    safe_target = target.replace("/", "_").replace(" ", "_")

    box_path = FIGURES_DIR / f"03_oblique_vs_transverse_box_strip_{safe_target}.png"
    ci_path = FIGURES_DIR / f"03_oblique_vs_transverse_mean_ci_{safe_target}.png"

    plot_oblique_vs_transverse_box_strip(
        df=df_pheno,
        target=target,
        result_row=row,
        save_path=box_path,
    )

    plot_oblique_vs_transverse_mean_ci(
        result_row=row,
        save_path=ci_path,
    )

print(f"Saved figures for {len(results_df)} targets to:")
print(FIGURES_DIR)
```
