```python
from pathlib import Path
import warnings

import numpy as np
import pandas as pd

import scipy.stats as stats
import statsmodels.formula.api as smf
from statsmodels.stats.diagnostic import het_breuschpagan

import seaborn as sns
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", 200)
pd.set_option("display.width", 200)

PROJECT_ROOT = Path(r"C:\Users\Ars\projects\university\lab_urfu_2026\false-lv-chords")

DATA_PATH = PROJECT_ROOT / "data" / "interim" / "regression_mechanistic_dataset.xlsx"

OUTPUT_DIR = PROJECT_ROOT / "output"
FIGURES_DIR = OUTPUT_DIR / "figures"
TABLES_DIR = OUTPUT_DIR / "tables"

FIGURES_DIR.mkdir(parents=True, exist_ok=True)
TABLES_DIR.mkdir(parents=True, exist_ok=True)

df = pd.read_excel(DATA_PATH)

print("Shape:", df.shape)
display(df.head())
```

```python
TARGETS = [
    "echo_lv_strain_plus_cv_pct",
    "echo_lv_strain_rate_cv_pct",
    "dicor_ssou_cv_pct",
    "dicor_sm_kdtr_cv_pct",
    "echo_lv_sphericity_index_end_systolic_ratio",
    "echo_lv_end_systolic_diameter_mm",
]

BINARY_PREDICTORS = [
    "transverse_any",
    "oblique_any",
    "long_oblique_bridge_any",
]

NUMERIC_PREDICTORS = [
    "transverse_load",
    "oblique_load",
    "long_oblique_bridge_load",
    "chord_type_diversity",
    "tethering_index",
]

CATEGORICAL_PREDICTORS = [
    "mechanical_chord_phenotype",
    "chord_type_complexity_class",
]

ALL_PREDICTORS = BINARY_PREDICTORS + NUMERIC_PREDICTORS + CATEGORICAL_PREDICTORS

TARGETS = [c for c in TARGETS if c in df.columns]
ALL_PREDICTORS = [c for c in ALL_PREDICTORS if c in df.columns]
BINARY_PREDICTORS = [c for c in BINARY_PREDICTORS if c in df.columns]
NUMERIC_PREDICTORS = [c for c in NUMERIC_PREDICTORS if c in df.columns]
CATEGORICAL_PREDICTORS = [c for c in CATEGORICAL_PREDICTORS if c in df.columns]

print("Targets:")
display(TARGETS)

print("Binary predictors:")
display(BINARY_PREDICTORS)

print("Numeric predictors:")
display(NUMERIC_PREDICTORS)

print("Categorical predictors:")
display(CATEGORICAL_PREDICTORS)
```

```python
df_model = df.copy()

for col in BINARY_PREDICTORS:
    df_model[col] = pd.to_numeric(df_model[col], errors="coerce")

for col in NUMERIC_PREDICTORS:
    df_model[col] = pd.to_numeric(df_model[col], errors="coerce")

for col in CATEGORICAL_PREDICTORS:
    df_model[col] = df_model[col].astype("object")

for col in TARGETS:
    df_model[col] = pd.to_numeric(df_model[col], errors="coerce")

display(df_model[TARGETS + ALL_PREDICTORS].head())
```

```python
def cohen_d_two_groups(y0, y1):
    y0 = pd.Series(y0).dropna()
    y1 = pd.Series(y1).dropna()
    
    n0 = len(y0)
    n1 = len(y1)
    
    if n0 < 2 or n1 < 2:
        return np.nan
    
    s0 = y0.std(ddof=1)
    s1 = y1.std(ddof=1)
    
    pooled_sd = np.sqrt(((n0 - 1) * s0**2 + (n1 - 1) * s1**2) / (n0 + n1 - 2))
    
    if pooled_sd == 0:
        return np.nan
    
    return (y1.mean() - y0.mean()) / pooled_sd


def get_robust_omnibus_p(model):
    """
    HC3-робастный omnibus-тест: все коэффициенты, кроме Intercept, равны 0.
    """
    try:
        robust = model.get_robustcov_results(cov_type="HC3")
        exog_names = model.model.exog_names
        
        non_intercept_idx = [i for i, name in enumerate(exog_names) if name != "Intercept"]
        
        if len(non_intercept_idx) == 0:
            return np.nan
        
        r_matrix = np.zeros((len(non_intercept_idx), len(exog_names)))
        
        for row_i, coef_i in enumerate(non_intercept_idx):
            r_matrix[row_i, coef_i] = 1
        
        f_test = robust.f_test(r_matrix)
        return float(f_test.pvalue)
    
    except Exception:
        return np.nan


def diagnostic_tests(model):
    residuals = model.resid
    
    try:
        shapiro_p = stats.shapiro(residuals).pvalue if 3 <= len(residuals) <= 5000 else np.nan
    except Exception:
        shapiro_p = np.nan
    
    try:
        bp_p = het_breuschpagan(residuals, model.model.exog)[1]
    except Exception:
        bp_p = np.nan
    
    try:
        influence = model.get_influence()
        cooks_d = influence.cooks_distance[0]
        max_cooks_d = float(np.nanmax(cooks_d))
        influential_n = int(np.sum(cooks_d > 4 / len(residuals)))
    except Exception:
        max_cooks_d = np.nan
        influential_n = np.nan
    
    return {
        "shapiro_resid_p": shapiro_p,
        "breusch_pagan_p": bp_p,
        "max_cooks_d": max_cooks_d,
        "influential_n_cook_4_over_n": influential_n,
    }
```

```python
def fit_main_effect_model(df, target, predictor, predictor_type, min_n=10, min_group_n=3):
    d = df[[target, predictor]].dropna().copy()
    
    if len(d) < min_n:
        return None
    
    if d[target].nunique(dropna=True) < 2:
        return None
    
    if d[predictor].nunique(dropna=True) < 2:
        return None
    
    group_counts = d[predictor].value_counts(dropna=False)
    min_observed_group_n = int(group_counts.min())
    
    if predictor_type in ["binary", "categorical"]:
        if min_observed_group_n < min_group_n:
            return None
        
        formula = f'Q("{target}") ~ C(Q("{predictor}"))'
    
    else:
        formula = f'Q("{target}") ~ Q("{predictor}")'
    
    model = smf.ols(formula, data=d).fit()
    
    diagnostics = diagnostic_tests(model)
    
    robust_omnibus_p = get_robust_omnibus_p(model)
    
    result = {
        "target": target,
        "predictor": predictor,
        "predictor_type": predictor_type,
        "n": int(model.nobs),
        "n_levels": int(d[predictor].nunique(dropna=True)),
        "min_group_n": min_observed_group_n if predictor_type in ["binary", "categorical"] else np.nan,
        "r2": float(model.rsquared),
        "adj_r2": float(model.rsquared_adj),
        "aic": float(model.aic),
        "bic": float(model.bic),
        "f_pvalue_classic": float(model.f_pvalue) if model.f_pvalue is not None else np.nan,
        "f_pvalue_hc3": robust_omnibus_p,
        "rmse": float(np.sqrt(np.mean(model.resid ** 2))),
        "mae": float(np.mean(np.abs(model.resid))),
        **diagnostics,
    }
    
    # Для числовых предикторов: beta, Pearson, Spearman
    if predictor_type == "numeric":
        term = f'Q("{predictor}")'
        
        robust = model.get_robustcov_results(cov_type="HC3")
        robust_params = pd.Series(robust.params, index=model.params.index)
        robust_pvalues = pd.Series(robust.pvalues, index=model.params.index)
        robust_conf = pd.DataFrame(
            robust.conf_int(),
            index=model.params.index,
            columns=["ci_low_hc3", "ci_high_hc3"]
        )
        
        try:
            pearson_r, pearson_p = stats.pearsonr(d[predictor], d[target])
        except Exception:
            pearson_r, pearson_p = np.nan, np.nan
        
        try:
            spearman_rho, spearman_p = stats.spearmanr(d[predictor], d[target])
        except Exception:
            spearman_rho, spearman_p = np.nan, np.nan
        
        result.update({
            "beta": float(model.params[term]),
            "beta_hc3": float(robust_params[term]),
            "beta_p_hc3": float(robust_pvalues[term]),
            "beta_ci_low_hc3": float(robust_conf.loc[term, "ci_low_hc3"]),
            "beta_ci_high_hc3": float(robust_conf.loc[term, "ci_high_hc3"]),
            "pearson_r": float(pearson_r),
            "pearson_p": float(pearson_p),
            "spearman_rho": float(spearman_rho),
            "spearman_p": float(spearman_p),
        })
    
    # Для бинарных предикторов: mean difference, Cohen's d, Mann-Whitney
    elif predictor_type == "binary":
        levels = sorted(d[predictor].dropna().unique())
        
        if len(levels) == 2:
            y0 = d.loc[d[predictor] == levels[0], target]
            y1 = d.loc[d[predictor] == levels[1], target]
            
            try:
                mw_p = stats.mannwhitneyu(y0, y1, alternative="two-sided").pvalue
            except Exception:
                mw_p = np.nan
            
            result.update({
                "group_0_label": str(levels[0]),
                "group_1_label": str(levels[1]),
                "group_0_n": int(len(y0)),
                "group_1_n": int(len(y1)),
                "group_0_mean": float(y0.mean()),
                "group_1_mean": float(y1.mean()),
                "mean_difference_1_minus_0": float(y1.mean() - y0.mean()),
                "cohen_d_1_minus_0": float(cohen_d_two_groups(y0, y1)),
                "mann_whitney_p": float(mw_p),
            })
    
    # Для категориальных предикторов: Kruskal-Wallis и разница крайних средних
    elif predictor_type == "categorical":
        grouped = [g[target].dropna() for _, g in d.groupby(predictor)]
        
        try:
            kruskal_p = stats.kruskal(*grouped).pvalue if len(grouped) >= 2 else np.nan
        except Exception:
            kruskal_p = np.nan
        
        group_stats = (
            d.groupby(predictor)[target]
            .agg(["count", "mean", "std", "median"])
            .reset_index()
            .sort_values("mean")
        )
        
        low_group = group_stats.iloc[0]
        high_group = group_stats.iloc[-1]
        
        result.update({
            "kruskal_p": float(kruskal_p),
            "lowest_mean_group": str(low_group[predictor]),
            "highest_mean_group": str(high_group[predictor]),
            "lowest_group_mean": float(low_group["mean"]),
            "highest_group_mean": float(high_group["mean"]),
            "highest_minus_lowest_mean": float(high_group["mean"] - low_group["mean"]),
        })
    
    return result
```

```python
rows = []

for target in TARGETS:
    for predictor in BINARY_PREDICTORS:
        result = fit_main_effect_model(
            df=df_model,
            target=target,
            predictor=predictor,
            predictor_type="binary",
            min_n=10,
            min_group_n=3,
        )
        if result is not None:
            rows.append(result)
    
    for predictor in NUMERIC_PREDICTORS:
        result = fit_main_effect_model(
            df=df_model,
            target=target,
            predictor=predictor,
            predictor_type="numeric",
            min_n=10,
            min_group_n=3,
        )
        if result is not None:
            rows.append(result)
    
    for predictor in CATEGORICAL_PREDICTORS:
        result = fit_main_effect_model(
            df=df_model,
            target=target,
            predictor=predictor,
            predictor_type="categorical",
            min_n=10,
            min_group_n=3,
        )
        if result is not None:
            rows.append(result)

screening = pd.DataFrame(rows)

if screening.empty:
    raise ValueError("No valid mechanistic main-effect models were fitted.")

screening["target_n_max"] = screening.groupby("target")["n"].transform("max")

# Логика отбора:
# 1) adjusted R² — насколько модель объясняет Y;
# 2) n — насколько модель держится на нормальной выборке;
# 3) для групповых моделей штрафуем слишком маленькие группы.
screening["n_factor"] = np.sqrt(screening["n"] / screening["target_n_max"])

screening["group_balance_factor"] = 1.0
mask_group = screening["predictor_type"].isin(["binary", "categorical"])
screening.loc[mask_group, "group_balance_factor"] = np.sqrt(
    np.minimum(screening.loc[mask_group, "min_group_n"] / 5, 1)
)

screening["selection_score"] = (
    screening["adj_r2"].clip(lower=0)
    * screening["n_factor"]
    * screening["group_balance_factor"]
)

screening = screening.sort_values(
    ["target", "selection_score", "adj_r2", "n"],
    ascending=[True, False, False, False]
)

display(screening)

screening.to_excel(TABLES_DIR / "02_mechanistic_main_effects_screening.xlsx", index=False)

print("Saved:")
print(TABLES_DIR / "02_mechanistic_main_effects_screening.xlsx")
```

```python
best_main_effects = (
    screening
    .sort_values(["target", "selection_score"], ascending=[True, False])
    .groupby("target")
    .head(1)
    .reset_index(drop=True)
)

display(best_main_effects[[
    "target",
    "predictor",
    "predictor_type",
    "n",
    "r2",
    "adj_r2",
    "selection_score",
    "f_pvalue_hc3",
    "f_pvalue_classic",
    "breusch_pagan_p",
    "shapiro_resid_p",
    "max_cooks_d",
    "influential_n_cook_4_over_n",
]])

best_main_effects.to_excel(TABLES_DIR / "02_best_mechanistic_main_effects.xlsx", index=False)

print("Saved:")
print(TABLES_DIR / "02_best_mechanistic_main_effects.xlsx")
```

```python
sns.set_theme(style="white", context="paper", font_scale=1.15)

def format_axes(ax):
    ax.set_facecolor("white")
    ax.grid(False)
    
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    
    ax.spines["left"].set_color("black")
    ax.spines["bottom"].set_color("black")
    ax.spines["left"].set_linewidth(1.1)
    ax.spines["bottom"].set_linewidth(1.1)
    
    ax.tick_params(axis="both", colors="black", width=1.0, length=4)
    ax.xaxis.label.set_color("black")
    ax.yaxis.label.set_color("black")


def prepare_order(d, predictor):
    values = d[predictor].dropna().unique().tolist()
    
    preferred_orders = {
        "transverse_any": [0, 1],
        "oblique_any": [0, 1],
        "long_oblique_bridge_any": [0, 1],
        "mechanical_chord_phenotype": [
            "No-selected-chord-types",
            "Oblique-Only",
            "Transverse-Involved",
        ],
        "chord_type_complexity_class": [
            "0_no_selected_types",
            "1_single_type",
            "2_two_types",
            "3_complex_network",
        ],
    }
    
    if predictor in preferred_orders:
        return [v for v in preferred_orders[predictor] if v in values]
    
    try:
        return sorted(values)
    except Exception:
        return values


def bin_numeric_predictor(series):
    def bin_value(x):
        if pd.isna(x):
            return np.nan
        if x == 0:
            return "0"
        if x == 1:
            return "1"
        if x == 2:
            return "2"
        return "≥3"
    
    return series.apply(bin_value)


def plot_box_strip_main_effect(df, target, predictor, predictor_type, model_row, save_path):
    d = df[[target, predictor]].dropna().copy()
    
    plot_predictor = predictor
    
    if predictor_type == "numeric":
        # Для счетчиков 0–3 показываем не scatter, а академический box+strip по категориям.
        d[f"{predictor}_bin"] = bin_numeric_predictor(d[predictor])
        plot_predictor = f"{predictor}_bin"
        order = [x for x in ["0", "1", "2", "≥3"] if x in d[plot_predictor].dropna().unique()]
    else:
        order = prepare_order(d, predictor)
    
    fig, ax = plt.subplots(figsize=(5.6, 4.4), dpi=300)
    
    sns.boxplot(
        data=d,
        x=plot_predictor,
        y=target,
        order=order,
        width=0.55,
        showfliers=False,
        ax=ax,
        boxprops={"facecolor": "white", "edgecolor": "black", "linewidth": 1.1},
        whiskerprops={"color": "black", "linewidth": 1.1},
        capprops={"color": "black", "linewidth": 1.1},
        medianprops={"color": "black", "linewidth": 1.2},
    )
    
    sns.stripplot(
        data=d,
        x=plot_predictor,
        y=target,
        order=order,
        ax=ax,
        color="black",
        size=4,
        jitter=0.18,
        alpha=0.65,
    )
    
    ax.set_xlabel(predictor)
    ax.set_ylabel(target)
    
    text = (
        f"n = {int(model_row['n'])}\n"
        f"R² = {model_row['r2']:.3f}\n"
        f"adj. R² = {model_row['adj_r2']:.3f}\n"
        f"p(HC3) = {model_row['f_pvalue_hc3']:.3f}"
    )
    
    ax.text(
        0.03,
        0.97,
        text,
        transform=ax.transAxes,
        ha="left",
        va="top",
        fontsize=9,
        color="black",
        bbox={"facecolor": "white", "edgecolor": "black", "linewidth": 0.6, "alpha": 0.9},
    )
    
    format_axes(ax)
    
    plt.xticks(rotation=20, ha="right")
    fig.tight_layout()
    fig.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.close(fig)
```

```python
top3_main_effects = (
    screening
    .sort_values(["target", "selection_score"], ascending=[True, False])
    .groupby("target")
    .head(3)
    .reset_index(drop=True)
)

display(top3_main_effects[[
    "target",
    "predictor",
    "predictor_type",
    "n",
    "r2",
    "adj_r2",
    "selection_score",
    "f_pvalue_hc3",
]])

for _, row in top3_main_effects.iterrows():
    target = row["target"]
    predictor = row["predictor"]
    predictor_type = row["predictor_type"]
    
    safe_target = target.replace("/", "_").replace(" ", "_")
    safe_predictor = predictor.replace("/", "_").replace(" ", "_")
    
    save_path = FIGURES_DIR / f"02_main_effect_{safe_target}__{safe_predictor}.png"
    
    plot_box_strip_main_effect(
        df=df_model,
        target=target,
        predictor=predictor,
        predictor_type=predictor_type,
        model_row=row,
        save_path=save_path,
    )

print(f"Saved {len(top3_main_effects)} figures to:")
print(FIGURES_DIR)
```
